```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from dotenv import load_dotenv
```

```python
load_dotenv()
embeddings = HuggingFaceEmbeddings(model_name='sentence-transformers/all-MiniLM-L6-v2')
```

```python
from langchain_core.documents import Document

docs1 = Document(
    page_content="Virat Kohli is one fo the most successful and consisent batsman in the IPL history. Known for his aggressive batting style...",
    meta_data={"team":"Royal Challengers Bangalore"}
)

docs2 = Document(
    page_content="MS Dhoni is one of the most iconic captains in IPL history. Known for his calm demeanor and finishing abilities, he has led Chennai Super Kings to multiple IPL titles...",
    meta_data={"team": "Chennai Super Kings"}
)

docs3 = Document(
    page_content="Rohit Sharma is a highly successful IPL captain and explosive opening batsman. He has led Mumbai Indians to several IPL championships and is known for his elegant stroke play...",
    meta_data={"team": "Mumbai Indians"}
)

docs4 = Document(
    page_content="Jasprit Bumrah is one of the best fast bowlers in IPL history. Known for his deadly yorkers and unique bowling action, he has been a key player for Mumbai Indians...",
    meta_data={"team": "Mumbai Indians"}
)

docs5 = Document(
    page_content="Hardik Pandya is one of the premier all-rounders in the IPL. Known for his explosive batting, medium-fast bowling, and dynamic fielding, he has played a crucial role in multiple IPL campaigns and has also captained Gujarat Titans to success...",
    meta_data={"team": "Mumbai Indians"}
)
```

```python
docs = [docs1, docs2, docs3, docs4, docs5]
```

```python
vector_store = Chroma(
    embedding_function=embeddings,
    persist_directory="Chroma_DB",
    collection_name="sample"
)
```

```python
vector_store.add_documents(docs)
```

```python
vector_store.get(include=['embeddings', 'documents', 'metadatas'])
```

```python
vector_store.similarity_search(
    query="Who among these are bowler?",
    k=2
)
```

```python
vector_store.similarity_search_with_score(
    query="Who among these are the best bowlers?",
    k=2
)
```

```python
vector_store.similarity_search_with_score(
    query="",
    filter={"team": "Mumbai Indians"}
)
```

```python
vector_store.get(include=["embeddings", "documents", "metadatas"])
```
